```python
import pandas as pd
import matplotlib.pyplot as plt

import numpy as np
```

```python
dataset =pd.read_csv("insurance.csv")
```

```python
dataset =pd.get_dummies(dataset,drop_first=True)
```

```python
dataset.columns
```

```python
independent = dataset[['age', 'bmi', 'children',  'sex_male', 'smoker_yes',
       'region_northwest', 'region_southeast', 'region_southwest']]
dependent = dataset[['expenses']]
                     
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(independent,dependent,test_size=0.2,random_state=10)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
from itertools import product
import pandas as pd

# Parameter values

n_estimators_values = [50, 100, 200]

criterion_values = [
    "squared_error",
    "absolute_error",
    "friedman_mse",
    "poisson"
]

max_depth_values = [None, 5, 10, 15, 20]

min_samples_split_values = [2, 5, 10]

min_samples_leaf_values = [1, 2, 4]

max_features_values = [1.0, "sqrt", "log2"]

bootstrap_values = [True, False]


# Store all results

results = []


# Test all combinations

for (
    n_estimators,
    criterion,
    max_depth,
    min_samples_split,
    min_samples_leaf,
    max_features,
    bootstrap
) in product(
    n_estimators_values,
    criterion_values,
    max_depth_values,
    min_samples_split_values,
    min_samples_leaf_values,
    max_features_values,
    bootstrap_values
):

    regressor = RandomForestRegressor(
        n_estimators=n_estimators,
        criterion=criterion,
        max_depth=max_depth,
        min_samples_split=min_samples_split,
        min_samples_leaf=min_samples_leaf,
        max_features=max_features,
        bootstrap=bootstrap,
        random_state=42,
        n_jobs=-1
    )

    regressor.fit(x_train, y_train)

    y_pred = regressor.predict(x_test)

    r2 = r2_score(y_test, y_pred)


    # Save results

    results.append({

        "n_estimators": n_estimators,
        "criterion": criterion,
        "max_depth": max_depth,
        "min_samples_split": min_samples_split,
        "min_samples_leaf": min_samples_leaf,
        "max_features": max_features,
        "bootstrap": bootstrap,
        "R2": r2

    })


# Convert results into DataFrame

results_df = pd.DataFrame(results)


# Sort by highest R²

results_df = results_df.sort_values(
    by="R2",
    ascending=False
)


# Display ALL combinations and R²

pd.set_option("display.max_rows", None)

print(results_df.to_string(index=False))


# Save to Excel

results_df.to_excel(
    "random_forest_all_results.xlsx",
    index=False
)


# Display best combination

print("\nBEST COMBINATION:")
print(results_df.iloc[0])
```
